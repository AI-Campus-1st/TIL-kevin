```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 한글 폰트 미리 설정하여 깨짐 방지
plt.rc('font', family='Malgun Gothic')
plt.rcParams['axes.unicode_minus']=False # 마이너스 깨짐 방지
```

```python
x = [10, 14, 19, 20, 25]
plt.plot(x)
plt.show()
```

```python
x = np.arange(-4.5, 5, 0.5) # (좌측끝, 우측끝, 간격)
y = x**2
plt.plot(x, y); plt.show()
```

```python
# 한 바닥에 동시에 그리기
y1 = x**2
y2 = x + 30
y3 = x**2 + 10
plt.plot(x, y1, label='y1') # 라벨 이름 지정
plt.plot(x, y2, color='#83ABF3') # 색깔 지정(hex code 또는 사전 지정된 프리셋 색깔 이름)
plt.plot(x, y3,linestyle='dashed',lw=4) # 선 스타일 지정

plt.axhline(20, color='gray',linestyle=':') # horizontal line
plt.axvline() # vertical line

plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = x**2

plt.plot(x, y, '--r')
plt.xlim(-20, 30) # x축 범위 제한
plt.ylim(-20, 120)
# plt.axis([-20, 30, -20, 120])
plt.xticks([-20,0,20]) # 축 눈금 설정
plt.yticks(np.arange(-20,120,10))

plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1, y2 = x**2, x**3

# 선 라벨
plt.plot(x, y1, 'r', label='line-1')
plt.plot(x, y2, 'g', label='line-2')

# 축 라벨
plt.xlabel('x-label'); plt.ylabel('y-label')

# plt.legend(loc=2) # 그래프 안
plt.legend(bbox_to_anchor=(1.04, 1)) #그래프 박스 밖
plt.text(4.5,200,'표시할 텍스트') # 그래프 박스 안(x,y,'text')

plt.show()
```
