```python
#1 재귀 함수를 사용했을 때 얻을 수 있는 장점과 단점을 반복문과 비교하여 작성하시오.
```

```python
# 일반 반복문
# 1*2*3*...*n

def fct(n):
    a=1
    for i in range(1,n+1):
        a*=i
        # print(a)
    return a

print(fct(5))
```

```python
# 재귀함수
# n! = n * (n-1)!

def fct(n):
    if n==1 or n==0:
        return 1
    return fct(n-1)*n

print(fct(5))
```

```python
#2 다음의 요구사항을 만족하는 Circle 클래스를 정의하고, 넓이와 둘레, 중심 좌표를 구하는 메서드를 정의하시오.

# 반지름: 3, x좌표: 2, y좌표: 4
class Circle:
    """원의 넓이, 둘레, 중심좌표를 구한다."""
    pi=3.14
    def __init__(self,r,x,y):
        self.r=r
        self.x=x
        self.y=y

    def area(self):
        print('%.2f'%(self.pi*self.r**2))

    def circumference(self):
        print('%.2f'%(self.pi*self.r*2))

    def center(self):
        print((self.x,self.y))

c1 = Circle(3, 2, 4)

c1.area() #=> 28.26
c1.circumference() #=> 18.84
c1.center() #=> (2, 4)
```
