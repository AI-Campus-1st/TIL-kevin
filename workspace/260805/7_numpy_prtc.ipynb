```python
import numpy as np
```

```python
matrix_1 = [
    [1, 2], 
    [3, 4]
]
matrix_2 = [
    [5, 6],
    [7, 8]
]
# 행렬 더하기
[
    [6, 8],
    [10, 12]
]

matrix_1 + matrix_2 # concat
result = []
for i in range(len(matrix_1)): # 행 인덱스
    temp = []
    for j in range(len(matrix_1[i])): # 열 인덱스
        # print(matrix_1[i][j], matrix_2[i][j])
        # temp에 더한 값을 append 해준다.
        temp.append(matrix_1[i][j] + matrix_2[i][j])
    # result에 temp(행 단위로 더한 리스트를) append한다.
    result.append(temp)
result
```

```python
# 모든 원자에 1씩 더하기
result = []
for i in range(len(matrix_1)): # 행 인덱스
    temp = []
    for j in range(len(matrix_1[i])): # 열 인덱스
        temp.append(matrix_1[i][j] + 1)
    result.append(temp)
result
```
