```python
import numpy as np
```

```python
sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가
```

브로드캐스트를 이용해 각 지점의 제품별 매출액(수량 × 단가) 행렬을 구하시오. (반복문 금지)

```python
sales * price # (3, 4) / (4, )
```

행렬곱을 이용해 각 지점의 총매출을 한 번에 구하시오. 또한 제품별 총판매량도 구하시오.


```python
# 행렬곱을 이용해 각 지점의 총매출
sales @ price

# 제품 별 총 판매량
sales.sum(axis=0)
```

```python
sum(sales[0] * price)
```

각 제품의 평균 판매량을 구한 뒤, 브로드캐스트로 평균 대비 편차 행렬을 구하시오.

```python
sales - sales.mean(axis=0)
```

각 지점을 기준으로 판매량을 정규화하시오. 결과의 행별 평균이 0, 표준편차가 1이 되는지 확인하시오.

```python
sales
```

```python
# 정규화
# (값 - 평균) / 표준편차

# 지점별 평균
# 지점별 표준편차
sales.mean(axis=1)
sales.std(axis=1)

            # (3, 4)    # (3, ) -> (3, 1)
sales_norm = (sales - sales.mean(axis=1).reshape(-1, 1)) / sales.std(axis=1).reshape(-1, 1)
sales_norm
```

```python
sales_norm.mean(axis=1)
sales_norm.std(axis=1)
```

```python
# 1차원에는 Transpose 불가
sales_norm = (sales.T - sales.mean(axis=1)) / sales.std(axis=1)
# sales_norm = sales_norm.T
```

```python
sales.T.shape
```

sales의 앞 3개 열만 잘라 3×3 정방행렬 A를 만들고, 행렬식과 역행렬을 구하시오. 

이어서 A와 역행렬을 곱해 단위행렬이 나오는지 확인하시오.

```python
A = sales[:, :3]
A

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)
```

```python
inv_A
```

```python
A @ inv_A
```

총매출이 가장 높은 지점의 인덱스(argmax)와, 지점 순서대로의 누적 매출(cumsum)을 구하시오.

```python
# 총매출이 가장 높은 지점의 인덱스(argmax)
total_sales = sales @ price
print(total_sales)
top_store_idx = total_sales.argmax()
print('가장 매출액이 높은 지점:', top_store_idx)

# 지점 순서대로의 누적 매출(cumsum)
total_sales.cumsum()
```
