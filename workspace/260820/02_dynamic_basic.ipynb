```python
URL = 'https://cine21.com/movie/rank/list_items/'


def fetch(page: int) -> dict:
    res = requests.get(URL, params={
        'p': page,
        'idx': 1 # 1주일, 1개월, 역대 (1, 2, 3)
    }, headers={
        'User-Agent': 'Jason/jason@fins.ai',
        "x-requested-with": "XMLHttpRequest",
        # referer
    })

    res.raise_for_status()
    return res.json()

def parse(data: dict) -> list[dict]:
    movies = []
    for item in data['items']:
        movies.append({
            '제목': item.get('hname', ''),
            '관객수': item.get('aud', -1),
            '누적관객수': item.get('aud_acc', -1),
            '썸네일': item.get('image', ''),
            'id': item.get('movie_id', '')
        })

    return movies

import time
movies = []
for page in range(1, 11):
    data = fetch(page)
    result = parse(data)

    movies.extend(result)
    time.sleep(0.5)

len(movies)
```

```python
# [m for m in movies if not m['id']]
```

```python
import pandas as pd
pd.DataFrame(movies).to_csv(
    "movie_ranking.csv", 
    index=False,
    encoding="utf-8-sig"
)
```

```python
URL = 'https://didimteo.startup-plus.kr/api/cportal/board'
# ?size=10&page=0&portalSeqNo=327&boardId=B000125&title=
PARAMS = {
    'size': 10,
    'portalSeqNo': 327,
    'boardId': 'B000125',
    'title': ''
}

def fetch(page: int) -> dict:
    res = requests.get(URL, 
                    params={**PARAMS, 'page': page}, 
                    headers={
                        'User-Agent': 'Mozilla 5.0' # Jason/jason@fins.ai
                        })

    res.raise_for_status()
    return res.json()

# JSON 포맷 안에 HTML 문서(데이터들 들어가있는)
# res.json()['html'] -> BeautifulSoup() 파싱 -> 선택자 써서 데이터 정리

def parse(data: dict) -> list[dict]:
    contents = []
    for item in data['data']['contents']:
        contents.append({
            '제목': item.get('title'),
            '작성일시': item.get('createDateTime'),
            '수정일시': item.get('updateDateTime'),
        })

    return contents

contents = []
for page in range(3):
    data = fetch(page)
    result = parse(data)

    contents.extend(result)
    time.sleep(0.5)

len(contents)
```
