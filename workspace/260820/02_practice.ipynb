```python
import requests

URL = 'https://ac.search.naver.com/nx/ac'

query_str = 'q=부트캠프&con=0&frm=nv&ans=2&r_format=json&r_enc=UTF-8&r_unicode=0&t_koreng=1&run=2&rev=4&q_enc=UTF-8&st=100&ackey=51tcli5j&_callback=_jsonp_7'

params = [q.split('=') for q in query_str.split('&')]
params = {k: v for k, v in params}
# %-encoding URL 같은거에 한글이나 특수기호같은거 넣을 때 사용하는 인코딩 형식

PARAMS = {
    'st': '100',
}

keyword = '인기영화'
res = requests.get(URL, params={**PARAMS, 'q': keyword})

res.status_code
```

```python
import json
body = res.text.split('(')[-1][:-1]

data = json.loads(body)

related_keywords = []
for item in data['items'][0]:
    related_keywords.append(item[0])
related_keywords
```

```python
# URL = 'https://ac.search.naver.com/nx/ac'
# PARAMS = {
#     'st': '100',
# }

def get_releated_keywords(keyword):
    res = requests.get(
        'https://ac.search.naver.com/nx/ac', 
        params={'st': '100', 'q': keyword})

    data = res.json()
    related_keywords = []
    for item in data['items'][0]:
        related_keywords.append(item[0])
    return related_keywords

get_releated_keywords('키보드')
```
