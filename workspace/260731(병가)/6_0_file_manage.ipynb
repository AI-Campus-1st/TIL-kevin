### 파일 관리

- `open('파일경로', 'mode')`: 파일을 여는 내장함수
    - mode: w(쓰기), r(읽기), a(수정) / wb, rb (바이너리 쓰기, 읽기)
    - encoding: 인코딩 설정 (utf-8)

```python
f = open('data.csv', 'r')
```

```python
# 한 줄 읽기 (커서가 이동)
columns = f.readline()

# 여러 줄 읽기 (현재 커서부터 끝까지)
lines = f.readlines()
```

```python
print(columns)
print(lines)
```

```python
columns = columns.rstrip().split(',')
```

```python
columns
```

```python
datas = []
for line in lines:
    data = line.rstrip().split(',')
    # 데이터 변환 로직 - 정수화
    data[1] = int(data[1])
    data[3] = int(data[3])
    datas.append(data)
datas
```

```python
f.close()
```

```python
columns, datas
```

```python
# 파일 저장 list[dict]

result = []
for data in datas:
    # print(dict(zip(columns, data)))
    temp = dict(zip(columns, data))
    result.append(temp)
# [
#     {'품목명': '초코우유', '주문개수': 4, '주문일시': '2026-07-31 11:03:10', '총액': 3200},
#     {'품목명': '매운새우깡', '주문개수': 2, '주문일시': '2026-06-30 10:02:35', '총액': 3000},
#     {'품목명': '육개장', '주문개수': 1, '주문일시': '2026-07-15 20:03:15', '총액': 1200},
# ]
```

### 데이터 분석 변환 & 저장

- 품목별 단가를 product.csv로 저장

```python
products = []
for data in result:
    name = data['품목명']
    unit_price = data['총액'] // data['주문개수']
    products.append({ '품목명':name, '단가': unit_price })
products
```

```python
# products
[
    {'품목명': '초코우유', '단가': 800},
]
```

```python
# product.csv에 저장
# 품목명,단가
# 초코우유,800
# 매운새우깡,1500
```

```python
# 품목명, 단가는 맨 위에 한번만 기록
# 데이터를 돌면서 값들만 기록
# 기록할 때에는 ,로 구분해서
```

```python
f = open('')
f.close()
```

```python
products
```

```python
# w - 새로 파일 생성
# a - 기존 파일 유지, 이어서 기록
with open('product.csv', 'w') as f:
    # 1. 파일에 뭔가를 쓴다 -> 텍스트 -> str
    # 2. 줄바꿈이 자동으로 들어가지 않음
    
    # 칼럼부터 작성
    f.write( ','.join(products[0].keys()) + '\n' )
    # f.write('하세요')
    # 데이터들을 작성
    for product in products:
        str_values = map(str, product.values())
        f.write( ','.join(str_values) + '\n' )

# with문을 벗어나면, 파일이 자동으로 닫힘
# f.write('작성 안되나요?')
```
