```python
#1 다음과 같이 Animal 클래스가 주어질 때, 해당 클래스를 상속 받아 아래의 보기와 같이 동작하는 Dog 클래스와 Bird 클래스를 작성하시오.

class Animal:
	def __init__(self, name):
		self.name = name
	
	def walk(self):
		print(f'{self.name}! 걷는다!')
	
	def eat(self):
		print(f'{self.name}! 먹는다!')
```

```python
class Dog(Animal):
    def __init__(self,name):
        super().__init__(name)

    def walk(self):
        print('%s! 달린다!'%self.name)
    def bark(self):
        print('%s! 짖는다!'%self.name)

class Bird(Animal):
    def __init__(self,name):
        super().__init__(name)

    def walk(self):
        print('%s! 걷는다!'%self.name)
    def eat(self):
        print('%s! 먹는다!'%self.name)
    def fly(self):
        print('%s! 푸드덕!'%self.name)
```

```python
dog = Dog('멍멍이')
dog.walk() # 멍멍이! 달린다!
dog.bark() # 멍멍이! 짖는다!

bird = Bird('구구')
bird.walk() # 구구! 걷는다!
bird.eat() # 구구! 먹는다!
bird.fly() # 구구! 푸드덕!
```

```python
#2 온라인 쇼핑몰의 결제 시스템을 상속 구조로 설계하세요.
'''
Payment 클래스

생성자에서 소유자 owner와 잔액 balance를 받아 저장한다.

수수료 메서드 (fee)를 정의하여 결제금액에 대한 수수료를 반환하는 추상메서드를 만든다. (Payment 클래스에서는 구현하지 않음)

결제 메서드 (pay)를 정의, 아래 규칙대로 구현하라

총 차감액 = amount + self.fee(amount)

잔액이 총 차감액보다 적으면 ValueError("잔액 부족")를 발생시킨다. (raise 활용)

충분하면 잔액에서 총 차감액을 빼고, {"amount": amount, "fee": 수수료, "balance": 결제후잔액} 딕셔너리를 반환한다.

결제수단 클래스

CardPayment: 수수료는 금액의 2.5% (소수점 버림, int)

MobilePayment: 수수료는 무조건 100원 정액

PointPayment: 수수료는 0원.
부모의 pay 메서드를 활용하고, {"point_used": True} 값을 추가해 반환한다.
'''
```

```python
class Payment:
    deduct=0
    def __init__(self,owner,balance):
        self.owner=owner
        self.balance=balance
    def fee(self,amount): # 추상 메서드
        '''Docstring'''
    def pay(self,amount):
        deduct=amount+self.fee(amount)
        if self.balance<deduct:
            raise ValueError("잔액 부족")
        self.balance-=deduct
        return {"amount": amount, "fee": self.fee(amount), "balance": self.balance}

class CardPayment(Payment):
    def fee(self,amount):
        return int(amount*0.025)
    

class MobilePayment(Payment):
    def fee(self,amount):
        return 100

class PointPayment(Payment):
    def fee(self,amount):
        return 0
    def pay(self, amount):
        result=super().pay(amount)
        result['point_used']=True
        return result
    
```

```python
print(CardPayment('kevin',34000).pay(12000))
```
