```python
from google import genai

prompt = """Please summarize the steps of cooking
and ingredients
(In Korean)

if there are more than one recipes, split it.

---
output format
[
    {
        "갈비찜": {{
            steps: [
                "1. ",
                "2. ",
                "3. ",
            ],
            ingredients: {
                "고춧가루": "1숟갈",
            }
        }}
    },
    ...
]
"""

client = genai.Client()
interaction = client.interactions.create(
    model='gemini-3.8-flash',
    input=[
        {"type": "text", "text": prompt},
        {
            "type": "video",
            "uri": "https://youtu.be/vM8uX9Pk-vE?si=1DAQkKTaaj_bG3-O"
        }
    ]
)
print(interaction.output_text)
```
