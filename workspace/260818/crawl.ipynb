```python
# 선택자 활용해서 HTML 문서의 파싱을 도와주는 라이브러리
%pip install beautifulsoup4
```

```python
from bs4 import BeautifulSoup

soup=BeautifulSoup(res.text,'html.parser')
soup
```

```python
soup.select('title')[0] # 특정 태그를 가져옴(list 형식이라 선택할 수 있음)
# 없으면 empty list 반환
```

```python
soup.select_one('#contentarea') # 하나만 가져오기
```

```python
soup.select('.no')[3] # 클래스가 number(.no)인 것들만 가져오기
```

```python
soup.find("title") # 가장 위의것 하나
soup.find("li", class_="type1") # class로 좁히기(class가 예약어라 _로 구분)
soup.find(id="contentarea") # id로 찾기

soup.find("a", attrs={"class":"thumb"}) # 없으면 None 반환
soup.find_all("li") # 조건에 맞는것 전부(리스트), 없으면 empty list 반환
```
