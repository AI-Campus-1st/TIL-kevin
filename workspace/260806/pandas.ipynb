```python
df=pd.read_csv('titanic.csv')
```

```python
df.describe() # include='O'(objects), 'all'(전체) (기본값 - 숫자만)
```

```python
# 1. 결측지 처리를 해주세요. 
# (age -평균값적용, Cabin -N으로, Embarked-N)

df.columns # columns가 많을 경우 수월한 확인을 위해
df['Age']=df['Age'].fillna(df['Age'].mean()) # 결측치에 대해 평균값으로 대입
df['Age'].isnull().sum() # 결측치 제거 확인
```

```python
df['Cabin']=df['Cabin'].fillna('N')
df['Embarked']=df['Embarked'].fillna('N')

df['Cabin'].isnull().sum()
df['Embarked'].isnull().sum()
```

```python
# 2. 성(Sex)별생존자 합을 구해주세요

# df[['Sex','Survived']]
df.groupby('Sex').sum()['Survived']
```

```python
# 3. 클래스(Pclass)별 생존자 합을 구해주세요

df.groupby('Pclass').sum()['Survived']
```

```python
# 4. 나이(Age)를 활용 유아, 10대, 20대, 30대, 40대, 50대, 60대, 노인의 생존자합을 구해주세요

def get_group(age):
    if age<10:
        return '영유아'
    
    for i in range(1,7):
        if age//10==i:
            return '%d0대'%i

    return '70대 이상'
```

```python
df['Age_group']=df['Age'].apply(get_group)
df.groupby('Age_group').sum()['Survived']
```
